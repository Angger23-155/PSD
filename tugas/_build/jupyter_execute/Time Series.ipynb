# **Time Series**

Time series adalah sekumpulan data yang dicatat atau diobservasi secara berurutan pada interval waktu tertentu. Data ini biasanya memiliki keteraturan temporal, artinya setiap pengamatan terjadi pada momen waktu yang spesifik, seperti harian, mingguan, bulanan, atau tahunan. Time series digunakan untuk menganalisis pola, tren, siklus, dan fluktuasi dalam data seiring waktu. Misalnya, data harga saham harian, jumlah penumpang kereta per bulan, atau curah hujan tahunan. Analisis time series membantu memprediksi perilaku di masa depan, mendeteksi anomali, dan memahami dinamika sistem yang berubah seiring waktu. Dalam praktiknya, time series sering melibatkan pemodelan statistik atau machine learning untuk membuat prediksi atau interpretasi yang lebih akurat.
