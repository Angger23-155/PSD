```python
import librosa
import numpy as np

def extract_features(file_path):
    y, sr = librosa.load(file_path, sr=None)

    features = []

    # --- 1️⃣ STATISTICAL FEATURES (5) ---
    features.extend([
        np.mean(y),
        np.std(y),
        np.max(y),
        np.min(y),
        np.ptp(y)
    ])

    # --- 2️⃣ TEMPORAL FEATURES (5) ---
    zcr = librosa.feature.zero_crossing_rate(y)
    rms = librosa.feature.rms(y=y)
    tempo, _ = librosa.beat.beat_track(y=y, sr=sr)

    features.extend([
        np.mean(zcr),
        np.std(zcr),
        np.mean(rms),
        np.std(rms),
        tempo
    ])

    # --- 3️⃣ SPECTRAL FEATURES (146 total) ---
    mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=20)          # 20 × 2 = 40
    chroma = librosa.feature.chroma_stft(y=y, sr=sr)            # 12 × 2 = 24
    contrast = librosa.feature.spectral_contrast(y=y, sr=sr)    # 7 × 2 = 14
    tonnetz = librosa.feature.tonnetz(y=librosa.effects.harmonic(y), sr=sr)  # 6 × 2 = 12
    centroid = librosa.feature.spectral_centroid(y=y, sr=sr)    # 1 × 2 = 2
    bandwidth = librosa.feature.spectral_bandwidth(y=y, sr=sr)  # 1 × 2 = 2
    rolloff = librosa.feature.spectral_rolloff(y=y, sr=sr)      # 1 × 2 = 2
```

```python
folder = "data"
data, labels = [], []

for label_dir in ["buka", "tutup"]:
    path_dir = os.path.join(folder, label_dir)
    for f in os.listdir(path_dir):
        if f.endswith(".wav"):
            file_path = os.path.join(path_dir, f)
            feats = extract_features(file_path)
            data.append(feats)
            labels.append(label_dir)

X = np.array(data)
y = np.array(labels)
print("Jumlah data:", len(y))
print("Jumlah fitur:", X.shape[1])
```
