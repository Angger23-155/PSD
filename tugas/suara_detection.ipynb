```python
import os
import librosa
import soundfile as sf
import numpy as np
from tqdm import tqdm

# === Folder dataset asli ===
folders = ['data/buka', 'data/tutup']

# === Fungsi augmentasi ===
def augment_audio(y, sr):
    augmented_audios = []

    # 1️⃣ Pitch shift (ubah nada suara)
    n_steps = np.random.uniform(-2, 2)  
    y_pitch = librosa.effects.pitch_shift(y, sr, n_steps)
    augmented_audios.append(y_pitch)

    # 2️⃣ Time stretch (ubah kecepatan bicara)
    rate = np.random.uniform(0.9, 1.1)
    y_stretch = librosa.effects.time_stretch(y, rate)
    augmented_audios.append(y_stretch)

    # 3️⃣ Tambah noise latar belakang
    noise = np.random.randn(len(y)) * 0.005
    y_noise = y + noise
    augmented_audios.append(y_noise)

    # 4️⃣ Kombinasi pitch + noise
    y_combo = librosa.effects.pitch_shift(y, sr, np.random.uniform(-1.5, 1.5))
    y_combo = y_combo + np.random.randn(len(y_combo)) * 0.004
    augmented_audios.append(y_combo)

    # 5️⃣ Time stretch + pitch
    y_mix = librosa.effects.time_stretch(y, np.random.uniform(0.95, 1.05))
    y_mix = librosa.effects.pitch_shift(y_mix, sr, np.random.uniform(-1, 1))
    augmented_audios.append(y_mix)

    return augmented_audios

# === Loop semua file dan augmentasi ===
for folder in folders:
    files = [f for f in os.listdir(folder) if f.endswith('.wav')]
    print(f"🔊 Memproses folder: {folder} (total {len(files)} file asli)")

    for f in tqdm(files):
        file_path = os.path.join(folder, f)
        y, sr = librosa.load(file_path, sr=None)

        # buat 5 variasi per file
        aug_audios = augment_audio(y, sr)

        # simpan hasil augmentasi
        base_name = os.path.splitext(f)[0]
        for i, aug in enumerate(aug_audios):
            out_name = f"{base_name}_aug{i+1}.wav"
            out_path = os.path.join(folder, out_name)
            sf.write(out_path, aug, sr)

print("✅ Augmentasi selesai! Sekarang setiap folder punya ±100 file suara.")
```
